```python
# Standardizing approved alternatives

def standardize_record(record):
    """
    Applies approved standardization mappings to a single record's
    market and unit fields. Returns the updated record.
    """
    # Standardize market name variants
    if record["market"] == "wakulima":
        record["market"] = "Wakulima"
    if record["market"] == "Nakuru top market":
        record["market"] = "Nakuru Top Market"

    # Standardize unit variants
    if record["unit"] in ("KG", "Kgs", "kilogram"):
        record["unit"] = "kg"
    if record["unit"] == "90 kg bag":
        record["unit"] = "90kg bag"

    #Stripping any whitespaces


    return record
```

```python
# Validating records

def validate_record(record):
    """
    Validates a single standardized record against the rules.
    Returns (True, None) if valid, or (False, reason) if invalid.
    """
    if not record["record_id"]:
        return False, "Missing Record ID"
    elif not record["market"]:
        return False, "Missing Market"
    elif not record["commodity"]:
        return False, "Missing Commodity"
    elif record["unit"] not in ("kg", "90kg bag"):
        return False, "Unsupported Unit"
    elif not isinstance(record["price_kes"], (int, float)) or record["price_kes"] <= 0:
        return False, "Invalid Price"
    elif not isinstance(record["quantity_available"], (int, float)) or record["quantity_available"] < 0:
        return False, "Invalid Quantity"
    else:
        return True, None
```

```python
def calculate_derived_fields(record):
    """
    Calculates price_per_kg and quantity_kg for a valid record
    and stores them in the record.
    """
    if record["unit"] == "kg":
        record["price_per_kg"] = record["price_kes"]
        record["quantity_kg"] = record["quantity_available"]
    else:  # "90kg bag"
        record["price_per_kg"] = record["price_kes"] / 90
        record["quantity_kg"] = record["quantity_available"] * 90
    return record
```

```python
def prepare_records():
    """
    Loads, standardizes, validates and calculates derived fields
    for all records. Returns (valid_records, invalid_records).
    """
    raw_records = load_data()
    valid_records = []
    invalid_records = []

    for record in raw_records:
        record = standardize_record(record)
        is_valid, reason = validate_record(record)

        if is_valid:
            record = calculate_derived_fields(record)
            valid_records.append(record)
        else:
            record["rejection_reason"] = reason
            invalid_records.append(record)

    return valid_records, invalid_records
```

```python
def get_unique_sorted_values(records, field):
    """
    Returns a sorted list of unique values for the given field
    across all records.
    """
    values = set()
    for r in records:
        values.add(r[field])
    return sorted(values)


def get_matching_records(records, commodity_name):
    """
    Returns all records whose commodity matches commodity_name
    (case-insensitive).
    """
    matches = []
    for r in records:
        if r["commodity"].lower() == commodity_name.lower():
            matches.append(r)
    return matches


def sort_records_by_price(records):
    """
    Returns a new list of records sorted by price_per_kg, ascending.
    """
    return sorted(records, key=lambda r: r["price_per_kg"])


def print_table(rows, headers, widths):
    """
    Prints rows as a simple fixed-width table using the given
    headers and column widths.
    """
    header_line = ""
    for header, width in zip(headers, widths):
        header_line += str(header).ljust(width)
    print(header_line)
    print("-" * sum(widths))

    for row in rows:
        line = ""
        for value, width in zip(row, widths):
            line += str(value).ljust(width)
        print(line)


def option_1_view_commodities(valid_records, invalid_records):
    commodities = get_unique_sorted_values(valid_records, "commodity")

    print("\n--- AVAILABLE COMMODITIES ---")
    for c in commodities:
        print("-", c)


def option_2_view_prices_for_commodity(valid_records, invalid_records):
    commodity_name = input("\nEnter commodity: ").strip()
    matches = get_matching_records(valid_records, commodity_name)

    if len(matches) == 0:
        print('No valid records found for "' + commodity_name + '".')
        return

    matches = sort_records_by_price(matches)

    print("\n--- " + commodity_name + " PRICES (lowest to highest) ---")
    headers = ["Market", "County", "Unit", "Price(KES)", "KES/kg"]
    widths = [20, 10, 10, 12, 8]
    rows = []
    for r in matches:
        rows.append((r["market"], r["county"], r["unit"], r["price_kes"], r["price_per_kg"]))
    print_table(rows, headers, widths)


def option_3_analyse_commodity(valid_records, invalid_records):
    commodity_name = input("\nEnter commodity to analyse: ").strip()
    matches = get_matching_records(valid_records, commodity_name)

    if len(matches) == 0:
        print('No valid records found for "' + commodity_name + '".')
        return

    # Find minimum and maximum by hand, without using min()/max()
    min_price = matches[0]["price_per_kg"]
    max_price = matches[0]["price_per_kg"]
    cheapest_market = matches[0]["market"]
    priciest_market = matches[0]["market"]

    for r in matches:
        if r["price_per_kg"] < min_price:
            min_price = r["price_per_kg"]
            cheapest_market = r["market"]
        if r["price_per_kg"] > max_price:
            max_price = r["price_per_kg"]
            priciest_market = r["market"]

    # Find average by hand
    total = 0
    for r in matches:
        total = total + r["price_per_kg"]
    avg_price = total / len(matches)

    above_avg_count = 0
    below_avg_count = 0
    for r in matches:
        if r["price_per_kg"] > avg_price:
            above_avg_count = above_avg_count + 1
        if r["price_per_kg"] < avg_price:
            below_avg_count = below_avg_count + 1

    print("\n--- ANALYSIS: " + commodity_name + " ---")
    print("Minimum price : KES " + str(round(min_price, 2)) + "/kg")
    print("Maximum price : KES " + str(round(max_price, 2)) + "/kg")
    print("Average price : KES " + str(round(avg_price, 2)) + "/kg")
    print("Cheapest market       : " + cheapest_market)
    print("Most expensive market : " + priciest_market)
    print("Records above average : " + str(above_avg_count))
    print("Records below average : " + str(below_avg_count))
```

Market comparison

```python
def calculate_price_per_kg(record):

    if record["unit"] == "kg":
        return record["price_kes"]

    elif record["unit"] == "90kg bag":
        return record["price_kes"] / 90
```

```python
def compare_markets(records, commodity, market1, market2):
    """
    Compares the price_per_kg of a given commodity
    between two specified markets.
    """
    market1_prices = []
    market2_prices = []

    for record in records:
        if record["commodity"].lower() == commodity.lower():
            if record["market"].lower() == market1.lower():
                market1_prices.append(record["price_per_kg"])
            elif record["market"].lower() == market2.lower():
                market2_prices.append(record["price_per_kg"])

    if not market1_prices:
        print(f"No price data for {commodity} in {market1}.")
        return
    if not market2_prices:
        print(f"No price data for {commodity} in {market2}.")
        return

    avg_price1 = sum(market1_prices) / len(market1_prices)
    avg_price2 = sum(market2_prices) / len(market2_prices)

    print(f"\n--- PRICE COMPARISON: {commodity} ---")
    print(f"Average price in {market1}: KES {avg_price1:.2f}/kg")
    print(f"Average price in {market2}: KES {avg_price2:.2f}/kg")

    if avg_price1 < avg_price2:
        print(f"{market1} has a lower average price for {commodity}.")
    elif avg_price2 < avg_price1:
        print(f"{market2} has a lower average price for {commodity}.")
    else:
        print(f"Prices for {commodity} are similar in {market1} and {market2}.")
```

```python
def option_4_compare_markets_input(valid_records):
    """
    Gets the commodity and two markets from the user,
    then compares their prices.
    """

    # Get available commodities
    commodities = []

    for record in valid_records:
        if record["commodity"] not in commodities:
            commodities.append(record["commodity"])

    commodities.sort()

    # Ask user for commodity
    commodity = input("\nEnter commodity to analyse market: ").strip()

    # Validate commodity
    selected_commodity = None

    for item in commodities:
        if item.lower() == commodity.lower():
            selected_commodity = item
            break

    if selected_commodity is None:
        print("Invalid commodity.")
        return

    # Get markets available for the selected commodity
    markets = []

    for record in valid_records:
        if record["commodity"].lower() == selected_commodity.lower():
            if record["market"] not in markets:
                markets.append(record["market"])

    markets.sort()

    # Display available markets
    print("\nAvailable markets:")
    for market in markets:
        print(f"- {market}") # market

    # Ask for first market
    market1 = input("\nEnter first market: ").strip()

    # Validate first market
    selected_market1 = None

    for market in markets:
        if market.lower() == market1.lower():
            selected_market1 = market
            break

    if selected_market1 is None:
        print("Invalid first market.")
        return

    # Ask for second market
    market2 = input("Enter second market: ").strip()

    # Validate second market
    selected_market2 = None

    for market in markets:
        if market.lower() == market2.lower():
            selected_market2 = market
            break

    if selected_market2 is None:
        print("Invalid second market.")
        return

    # Make sure the user does not select the same market twice
    if selected_market1.lower() == selected_market2.lower():
        print("Please select two different markets.")
        return

    # Call the comparison function
    compare_markets(
        valid_records,
        selected_commodity,
        selected_market1,
        selected_market2
    )
```

```python
valid_records, invalid_records = prepare_records()

option_1_view_commodities(valid_records, invalid_records)
option_2_view_prices_for_commodity(valid_records, invalid_records)
option_3_analyse_commodity(valid_records, invalid_records)
option_4_compare_markets_input(valid_records)
```
