```python
# Standardizing approved alternatives

def standardize_record(record):
    """
    Applies approved standardization mappings to a single record's
    market and unit fields. Returns the updated record.
    """
    # Standardize market name variants
    if record["market"] == "wakulima":
        record["market"] = "Wakulima"
    if record["market"] == "Nakuru top market":
        record["market"] = "Nakuru Top Market"

    # Standardize unit variants
    if record["unit"] in ("KG", "Kgs", "kilogram"):
        record["unit"] = "kg"
    if record["unit"] == "90 kg bag":
        record["unit"] = "90kg bag"

    #Stripping any whitespaces
    

    return record
```

```python
# Validating records

def validate_record(record):
    """
    Validates a single standardized record against the rules.
    Returns (True, None) if valid, or (False, reason) if invalid.
    """
    if not record["record_id"]:
        return False, "Missing Record ID"
    elif not record["market"]:
        return False, "Missing Market"
    elif not record["commodity"]:
        return False, "Missing Commodity"
    elif record["unit"] not in ("kg", "90kg bag"):
        return False, "Unsupported Unit"
    elif not isinstance(record["price_kes"], (int, float)) or record["price_kes"] <= 0:
        return False, "Invalid Price"
    elif not isinstance(record["quantity_available"], (int, float)) or record["quantity_available"] < 0:
        return False, "Invalid Quantity"
    else:
        return True, None
```

```python
def calculate_derived_fields(record):
    """
    Calculates price_per_kg and quantity_kg for a valid record
    and stores them in the record.
    """
    if record["unit"] == "kg":
        record["price_per_kg"] = record["price_kes"]
        record["quantity_kg"] = record["quantity_available"]
    else:  # "90kg bag"
        record["price_per_kg"] = record["price_kes"] / 90
        record["quantity_kg"] = record["quantity_available"] * 90
    return record
```

```python
def prepare_records():
    """
    Loads, standardizes, validates and calculates derived fields
    for all records. Returns (valid_records, invalid_records).
    """
    raw_records = load_data()
    valid_records = []
    invalid_records = []

    for record in raw_records:
        record = standardize_record(record)
        is_valid, reason = validate_record(record)

        if is_valid:
            record = calculate_derived_fields(record)
            valid_records.append(record)
        else:
            record["rejection_reason"] = reason
            invalid_records.append(record)

    return valid_records, invalid_records
```
